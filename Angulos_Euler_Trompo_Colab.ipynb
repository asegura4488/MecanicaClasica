# Ángulos de Euler: un trompo en 3D

Este cuaderno está preparado para **Google Colab** y permite presentar las tres rotaciones de un sólido rígido.

- **φ (phi) — precesión:** rotación alrededor del eje fijo $z$.
- **θ (theta) — nutación:** inclinación del eje de simetría del trompo con respecto al eje fijo $z$.
- **ψ (psi) — giro propio:** rotación del trompo sobre su eje de simetría $z'$.

Usaremos la **convención ZXZ** de rotaciones activas:

$$oxed{R(\phi,	heta,\psi)=R_z(\phi)R_x(	heta)R_z(\psi).}$$

La punta del trompo permanece en el origen. Como el sólido es axialmente simétrico, tiene una **franja naranja** para distinguir visualmente la rotación $\psi$.

## 2. ¿De dónde sale la rotación?

Si el punto de un sólido, en las coordenadas ligadas al propio cuerpo, es

$$\mathbf r_{\mathrm{cuerpo}}=\begin{pmatrix}x\\y\\z\end{pmatrix},$$

la posición del punto vista desde el laboratorio se calcula mediante

$$\mathbf r_{\mathrm{mundo}}=R\,\mathbf r_{\mathrm{cuerpo}},\qquad R=R_z(\phi)R_x(\theta)R_z(\psi).$$

Las matrices de rotación elementales son

$$R_z(\alpha)=\begin{pmatrix}\cos\alpha&-\sin\alpha&0\\\sin\alpha&\cos\alpha&0\\0&0&1\end{pmatrix},\qquad
R_x(\beta)=\begin{pmatrix}1&0&0\\0&\cos\beta&-\sin\beta\\0&\sin\beta&\cos\beta\end{pmatrix}.$$

**Importante:** al multiplicar matrices, la matriz de la derecha actúa primero en el vector. La rotación $\psi$ ocurre alrededor del eje original de simetría del cuerpo, después se aplica la inclinación $\theta$ y finalmente la precesión $\phi$ respecto al eje fijo.

El eje de simetría del trompo tiene vector director $\hat{\mathbf z}'=R(0,0,1)^T$. Calculando el producto:

$$\boxed{\hat{\mathbf z}'=(\sin\theta\sin\phi,\,-\sin\theta\cos\phi,\,\cos\theta).}$$

Este vector es **independiente de $\psi$** porque girar un sólido alrededor de su propio eje no modifica la orientación de ese eje. Por eso necesitamos dibujar una marca material sobre el trompo.

### Dos actividades para la clase

1. Fija $\theta=45^\circ$ y $\psi=0^\circ$; cambia solamente $\phi$. Observa el cono descrito por el eje de simetría.
2. Fija $\phi$ y $\theta$, y cambia solamente $\psi$. Observa cómo gira la marca naranja mientras el eje $z'$ permanece inmóvil.

**Caso especial:** cuando $\theta=0$, las rotaciones $\phi$ y $\psi$ son alrededor del mismo eje. En esta orientación no se pueden distinguir de forma única: es una singularidad de las coordenadas de Euler (bloqueo de cardán, *gimbal lock*).

```python
import numpy as np

def Rz(angulo):
    c, s = np.cos(angulo), np.sin(angulo)
    return np.array([[c,-s,0],[s,c,0],[0,0,1]])

def Rx(angulo):
    c, s = np.cos(angulo), np.sin(angulo)
    return np.array([[1,0,0],[0,c,-s],[0,s,c]])

phi, theta, psi = np.deg2rad([40, 45, 20])
R = Rz(phi) @ Rx(theta) @ Rz(psi)
print('Matriz R =\n', np.round(R, 3))
print('\nEje z prima =', np.round(R @ [0,0,1], 3))
print('\nR.T @ R =\n', np.round(R.T @ R, 10))
print('Determinante =', round(np.linalg.det(R), 10))
```

## 3. Versión completamente en Python (opcional)

Esta alternativa utiliza **NumPy + Matplotlib + ipywidgets**. Es útil para explicar el algoritmo paso a paso y modificar el perfil del trompo desde Python. Los cambios se recalculan en el kernel y por ello **son menos fluidos** que el visualizador anterior. Para la demostración rápida en clase recomiendo la primera versión.

```python
# Ángulos de Euler: trompo interactivo con Python, NumPy y Matplotlib.
# Compatible con Google Colab. Ejecuta el archivo en una celda de Colab,
# o copia todo su contenido a una celda.
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


def Rz(angulo):
    """Rotación alrededor del eje Z (ángulo en radianes)."""
    c, s = np.cos(angulo), np.sin(angulo)
    return np.array([[c, -s, 0], [s, c, 0], [0, 0, 1]])


def Rx(angulo):
    """Rotación alrededor del eje X (ángulo en radianes)."""
    c, s = np.cos(angulo), np.sin(angulo)
    return np.array([[1, 0, 0], [0, c, -s], [0, s, c]])


def rotacion_euler(phi, theta, psi):
    """Convención ZXZ: Rz(phi) @ Rx(theta) @ Rz(psi), grados."""
    p, t, s = np.deg2rad([phi, theta, psi])
    return Rz(p) @ Rx(t) @ Rz(s)


# Perfil del trompo: altura desde la punta / radio del cuerpo.
alturas = np.array([0, .10, .22, .42, .70, 1.01, 1.24, 1.40,
                    1.67, 1.92, 2.12, 2.30, 2.55, 2.63, 2.68])
radios = np.array([0, .09, .20, .38, .69, .96, 1.08, 1.05,
                   .86, .56, .27, .17, .17, .12, 0])
z = np.linspace(0, 2.68, 36)
a = np.linspace(0, 2*np.pi, 52)
Z, A = np.meshgrid(z, a, indexing='ij')
RAD = np.interp(Z, alturas, radios)
# Superficie del sólido en el sistema de coordenadas del propio trompo.
P = np.array([RAD*np.cos(A), RAD*np.sin(A), Z])


def mostrar_trompo(phi=40, theta=40, psi=20):
    R = rotacion_euler(phi, theta, psi)
    # Rotación rígida de cada punto: p_mundo = R @ p_cuerpo.
    X, Y, Zr = (R @ P.reshape(3, -1)).reshape(P.shape)
    fig = plt.figure(figsize=(7, 6))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Zr, color='#4688b0', alpha=.94,
                    linewidth=0, antialiased=True, shade=True)

    # Marca naranja sobre la superficie: permite ver el giro psi.
    rr = np.interp(z, alturas, radios) + .025
    linea = R @ np.array([rr, np.zeros_like(z), z])
    ax.plot(*linea, color='#f39c29', linewidth=4, label='Marca del cuerpo')

    # Ejes de coordenadas fijos en el espacio.
    L = 3.0
    for vec, color, etiqueta in [([1,0,0], '#c45b5b', 'x'),
                                 ([0,1,0], '#5a9a69', 'y'),
                                 ([0,0,1], '#6d758b', 'z')]:
        vec = L * np.array(vec)
        ax.plot([0,vec[0]], [0,vec[1]], [0,vec[2]], color=color, lw=2)
        ax.text(*vec, etiqueta, fontsize=12, weight='bold', color=color)

    # Eje de simetría del trompo, expresado en coordenadas del espacio.
    eje = R @ np.array([0, 0, 3.05])
    ax.plot([0,eje[0]], [0,eje[1]], [0,eje[2]],
            color='#008f82', lw=3, label="Eje z' del trompo")
    ax.scatter(*eje, s=38, color='#008f82')

    # Circunferencia que recorre el extremo de z' al variar phi.
    ang = np.linspace(0, 2*np.pi, 100)
    t = np.deg2rad(theta)
    rho, h = 3.05*np.sin(t), 3.05*np.cos(t)
    ax.plot(rho*np.sin(ang), -rho*np.cos(ang), h*np.ones_like(ang),
            '--', color='#a78cbd', lw=1.5, label='Precesión')

    ax.set_xlim(-3.4, 3.4)
    ax.set_ylim(-3.4, 3.4)
    ax.set_zlim(-3.4, 3.4)
    ax.set_box_aspect((1,1,1))
    ax.set_xlabel('X fijo')
    ax.set_ylabel('Y fijo')
    ax.set_zlabel('Z fijo')
    ax.view_init(elev=23, azim=-60)
    ax.set_title(f'φ = {phi:.0f}°,  θ = {theta:.0f}°,  ψ = {psi:.0f}°')
    ax.legend(loc='upper left', fontsize=8)
    plt.tight_layout()
    plt.show()
    plt.close(fig)


# Deslizadores conectados a la función que dibuja el sólido.
phi_slider = widgets.IntSlider(value=40, min=0, max=360, step=1,
                               description='φ (precesión)', continuous_update=True,
                               style={'description_width':'initial'})
theta_slider = widgets.IntSlider(value=40, min=0, max=180, step=1,
                                 description='θ (nutación)', continuous_update=True,
                                 style={'description_width':'initial'})
psi_slider = widgets.IntSlider(value=20, min=0, max=360, step=1,
                               description='ψ (giro propio)', continuous_update=True,
                               style={'description_width':'initial'})

interactivo = widgets.interactive_output(mostrar_trompo,
    {'phi': phi_slider, 'theta': theta_slider, 'psi': psi_slider})
display(widgets.VBox([phi_slider, theta_slider, psi_slider]), interactivo)
```
